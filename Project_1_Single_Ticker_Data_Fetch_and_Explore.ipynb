```python
import pandas as pd
import numpy as np
import requests
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
ticker = yf.Ticker("GC=F")
data = ticker.history(period='1y')
```

```python
data.tail(5)
```

```python
data.columns
```

```python
data.shape
```

```python
#Trying to get a single column (returns a series)
data['Close']
```

```python
print(data['Close'].iloc[0])        #First Close Value
print(data['Close'].iloc[-1])       #Last Close Value
```

```python
print(data['Close'].min())
print(data['Close'].max())
print(data['Close'].mean())
```

```python
print(((data['Close'].iloc[-1] - data['Close'].iloc[0]) * 100) /  data['Close'].iloc[-1])   #Total Return over the year
```

```python
plt.figure(figsize=(10, 5))
plt.plot(data.index, data['Close'])
plt.title('Gold Futures Closing Price Last Year')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.grid(True)
plt.show()
```
